```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_excel('/content/Sample data (1).xlsx')

df.head()
```

```python
print("Dataset Shape:", df.shape)

print("\nColumn Names:")
print(df.columns)

print("\nDataset Information:")
df.info()
```

```python
print("Missing Values:")
print(df.isnull().sum())
```

```python
df['Discount Band'] = df['Discount Band'].fillna('Unknown')

print(df['Discount Band'].isnull().sum())

```

```python
print("Duplicate rows:", df.duplicated().sum())
```

```python
df.describe()
```

```python
product_sales = df.groupby('Product')['Sales'].sum()

print(product_sales.sort_values(ascending=False))
```

```python
print(df.columns.tolist())
```

```python
df.columns = df.columns.str.strip()

print(df.columns.tolist())
```

```python
product_sales = df.groupby('Product')['Sales'].sum()

print(product_sales.sort_values(ascending=False))
```

```python
product_profit = df.groupby('Product')['Profit'].sum()

print(product_profit.sort_values(ascending=False))
```

```python
country_sales = df.groupby('Country')['Sales'].sum()

print(country_sales.sort_values(ascending=False))
```

```python
country_sales.sort_values().plot(
    kind='bar',
    title='Total Sales by Country',
    figsize=(8,5)
)

plt.xlabel('Country')
plt.ylabel('Total Sales')
plt.xticks(rotation=45)
plt.show()
```

```python
product_sales.sort_values().plot(
    kind='bar',
    title='Total Sales by Product',
    figsize=(8,5)
)

plt.xlabel('Product')
plt.ylabel('Total Sales')
plt.xticks(rotation=45)
plt.show()
```

```python
product_profit.sort_values().plot(
    kind='bar',
    title='Total Profit by Product',
    figsize=(8,5)
)

plt.xlabel('Product')
plt.ylabel('Total Profit')
plt.xticks(rotation=45)
plt.show()
```

```python
yearly_sales = df.groupby('Year')['Sales'].sum()

print(yearly_sales)
```

```python
yearly_sales.plot(
    kind='bar',
    title='Yearly Sales Analysis',
    figsize=(7,5)
)

plt.xlabel('Year')
plt.ylabel('Total Sales')
plt.xticks(rotation=0)
plt.show()
```

```python
print("SALES DATA ANALYSIS SUMMARY")
print("----------------------------")

print("Total Sales:", df['Sales'].sum())
print("Total Profit:", df['Profit'].sum())
print("Total Units Sold:", df['Units Sold'].sum())

print("\nBest Selling Product:", product_sales.idxmax())
print("Most Profitable Product:", product_profit.idxmax())
print("Top Sales Country:", country_sales.idxmax())
print("Total Products:", df['Product'].nunique())
print("Total Countries:", df['Country'].nunique())
```
